# RL-ASL 4-PLD Clinical Inference
This notebook loads the trained 4-PLD SAC model and runs inference on real clinical NIfTI datasets, displaying the results directly in the notebook.


```python
%matplotlib inline
from pathlib import Path
import torch
from rl_4pld_validation import clinical_preprocess, load_agent

BASE = Path.cwd()
DATA_ROOT = BASE / 'test data'
OUTPUT_ROOT = BASE / 'rl_inference_output'
CHECKPOINT = BASE / 'experiments' / 'rl_sac_4pld_trained' / 'notebook_results' / 'sac_asl_final.pth'
OUTPUT_ROOT.mkdir(exist_ok=True)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
agent, checkpoint_report = load_agent(CHECKPOINT, device)
clinical_report, clinical_data = clinical_preprocess(DATA_ROOT)
print('Using device:', device)
print('Checkpoint validation:', 'PASS' if checkpoint_report['keys_match'] else 'FAIL')
print('Actor mode: evaluation')
print('Actions: deterministic')
print('Optimizer called: False')
print('Clinical acquisition status:', clinical_report['status'])
print('Valid voxels:', clinical_report.get('valid_voxels'))
if clinical_report['status'] != 'VERIFIED_ACQUISITION_ORDER':
    print('STOP: clinical maps are not generated until the four volumes are mapped to the training PLD/LD table.')
    print(clinical_report['message'])
```

```python
if clinical_data is not None and clinical_report['status'] == 'VERIFIED_ACQUISITION_ORDER':
    print('Clinical rollout is enabled only after acquisition-order verification.')
else:
    print('No CBF_RL.nii.gz or ATT_RL.nii.gz written: acquisition order is ambiguous.')
    print('Preprocessing diagnostics are available in:', OUTPUT_ROOT / 'inference_summary.json')
```
